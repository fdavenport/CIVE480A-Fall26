**Background**

You are an engineer on a committee that is updating the wind design requirements for buildings in Mobile, Alabama, a city on the Gulf Coast that is frequently affected by hurricanes. 

In the United States, *ASCE 7* (Minimum Design Loads and Associated Criteria for Buildings and Other Structures) defines the design wind speed as the wind speed with a specific return period. The return period  used depends on the **Risk Category** of the building, which reflects the consequences of a failure:

| Risk Category | Examples | ASCE 7-16 return period |
|---|---|---|
| I | Buildings that pose a low risk to human life if they fail (agricultural buildings, minor storage buildings) | 300 years |
| II | Most ordinary buildings (houses, offices) | 700 years |
| III | Buildings where failure would pose a substantial risk to human life (schools, large assembly buildings, etc.) | 1,700 years |
| IV | Essential facilities (hospitals, fire stations, emergency shelters, etc.) | 3,000 years |

Buildings constructed today are expected to last 50-100 years or more, so the climate that they will experience may be different from the historical climate. **Your committee's job is to evaluate whether the current design wind speeds are still adequate in a future climate.**

**Fitting a GEV distribution**

Block maxima (like our 20-year maximum wind speeds) are described by the Generalized Extreme Value (GEV) distribution. The GEV distribution has three parameters: the location, the scale, and the shape. We will use the `genextreme.fit()` function from the scipy package to estimate the parameters of the GEV distribution that best fit the wind speed data. The function returns the **shape**, **location**, and **scale** (in that order).

The next cell fits a GEV distribution to the 2015 data:

```python
shape_2015, location_2015, scale_2015 = genextreme.fit(wind_2015["max_wind"])

print("shape:", shape_2015)
print("location:", location_2015)
print("scale:", scale_2015)
```

**Calculating return periods**

The `genextreme.isf()` function calculates the event size for a given exceedance probability. (`isf` stands for Inverse Survival Function). The first argument to `genextreme.isf()` is the exceedance probability, followed by the shape, location, and scale parameters.

Our GEV distribution describes the maximum wind speed in a 20-year block, so we need to convert the return period (in years) to the probability that the wind speed is exceeded in a 20-year block. For example, a 100-year event has a 1% probability of being exceeded in a single year, but a $20/100 = 0.2$ (or 20%) chance of being exceeded in a 20-year block. We can convert the annual return periods to the probability of exceedance in a 20-year block using $20/T$, where T is the return period in years. 

This only works for return periods that are longer than the block (20 years), so we will only look at return periods of 50 years or longer.

```python
block_length = 20 ## length of each block in years

return_periods = np.array([50, 100, 300, 500, 700, 1000, 1700, 3000])
exceedance_prob = block_length / return_periods ## exceedance probability within 20-year block of time
```

The next cell calculates the wind speed for each return period in the 2015 simulation and prints the results in a table:

```python
wind_return_periods_2015 = genextreme.isf(exceedance_prob, shape_2015, location_2015, scale_2015)

pd.DataFrame({"return period (years)": return_periods, "2015 wind speed (m/s)": wind_return_periods_2015.round(1)})
```

**Comparing the GEV distribution to the data**

We can compare the GEV return levels to the *empirical* return periods of the data. The empirical return period of each data point is based on how often values in our data were exceeded within the 20,000 simulated years. For example, with 970 20-year blocks, there are 19,400 years of data. The largest value was exceeded once in 19,400 years, so its empirical return period is approximately ~19,400 years. The second largest value was exceeded twice in 19,400 years, so its empirical return period is approximatey 19,400 / 2 = ~9,700 years. 

The function below calculates the empirical return period of each value in the data (in units of blocks).

```python
# run this code cell
def calc_empirical_return_level(data):
    """
    Compute empirical return level
    """
    df = pd.DataFrame(index=np.arange(data.size))
    # sort the data
    df["sorted_value"] = np.sort(data)[::-1]
    # rank via scipy instead to deal with duplicate values
    df["ranked_value"] = np.sort(stats.rankdata(-data))
    # find exceedance probability
    n = data.size
    df["exceedance"] = df["ranked_value"] / (n + 1)
    # find return period (in number of blocks)
    df["return_period"] = 1 / df["exceedance"]

    df = df[::-1]

    return df
```
