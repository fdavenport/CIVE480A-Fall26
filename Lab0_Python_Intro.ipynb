### Creating a list and indexing it

A **list** stores multiple values together in a single variable, in a specific order. Create one with square brackets `[]`, separating values with commas. You can access a single value in the list using its **index** (its position), also with square brackets. Python indexing starts at 0, so the first element is at index `0`, and you can use `-1` to get the last element.

```python
# create a list of daily rainfall totals (mm) for one week
daily_rainfall = [12, 8, 0, 24, 5, 30, 2]

daily_rainfall
```

```python
# index the first day's rainfall (python starts counting at zero)
print("first day's rainfall:", daily_rainfall[0])

# you can index from the end of the list using negative indices. [-1] selects the last day's rainfall
print("last day's rainfall:", daily_rainfall[-1])

# a "slice" selects a range of elements (this selects up to but not including the second number in the slice)
print("three days of rainfall:", daily_rainfall[0:3])
```

### Loops

A **for loop** lets us repeat an action once for each value in a list, instead of writing the same code multiple times.

The commands to create a "for loop" must be written in a certain way (called "syntax")

```python
my_list = [1, 2, 3, 4, 5] # here is a basic list to demonstrate a for loop 

# the first line of the for loop tells what should be looped through (in this case, my_list)
# and what each element of the list will be stored as within the loop (in this case, "x")
for x in my_list: 
    print(x)
    # code that should run within the loop must be indented 4 spaces
   
```

```python
# a basic for loop using our daily_rainfall list: 
for rainfall in daily_rainfall:
    print("daily rainfall:", rainfall)
```

```python
# combine the loop with an if statement to check a condition for each day
# here, we flag any day with more than 20 mm of rainfall as a "heavy rain" day
for rainfall in daily_rainfall:
    if rainfall > 20:
        print(rainfall, "mm - heavy rain day!")
        # code that should run if the if statement is True must be indented 4 spaces
    else:
        print(rainfall)
        # code that will run if the if statement is False
```

### Writing your own function

### Making a basic figure

`matplotlib` is a widely used library for making plots. We'll import its `pyplot` module (nicknamed `plt`) and use `plt.plot()` to make a simple line plot.

```python
import matplotlib.pyplot as plt

# plot the daily rainfall totals from earlier
# first we need to make a list of days to use as the x values for our plot
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, daily_rainfall) # the arguments for plt.plot are (x, y)
plt.xlabel("Day")
plt.ylabel("Rainfall (mm)")
plt.title("Daily Rainfall")

plt.show()
```

### Formatting a figure

Plots can be customized in many ways: `color` and `marker` change how a line or set of points looks, `label` (together with `plt.legend()`) identifies each line in a legend, and `fontsize` controls the size of labels, titles, and legend text. You can also add extra reference lines, like `plt.axhline()`, which draws a horizontal line at a given y-value.

```python
# plot the daily rainfall, plus a reference line marking the "heavy rain" threshold
plt.plot(days, daily_rainfall, color="blue", marker="o", label="Daily rainfall")
plt.axhline(y=20, color="red", linestyle="--", label="Heavy rain threshold")

plt.xlabel("Day", fontsize=12)
plt.ylabel("Rainfall (mm)", fontsize=12)
plt.title("Daily Rainfall", fontsize=14)
plt.legend()

plt.show()
```
