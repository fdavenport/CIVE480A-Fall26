# Part 1: Increases in Atmospheric Water Vapor
As we discussed in class, the saturation vapor pressure of air increases with temperature, as described by the Clausius-Clapeyron relation. The following equation can be used to calculate the saturation vapor pressure of water for a given temperature:

$\Large e_s = 0.611 exp(\frac{L_v}{R_v}(\frac{1}{273.16}-\frac{1}{T})$

where 
* $e_s$ is the saturation vapor pressure in kPa
* $L_v = 2.5 x 10^6 J/kg$ is the latent heat of vaporization for water
* $R_v = 461 J/kg*K$ is the ideal gas constant for water vapor, and
* $T$ is the temperature in $K$.

**Calculating the maximum SWE measurement in a year**

Water managers often use the maximum amount of Snow Water Equivalent each winter as an indication of how much streamflow will be available over the coming spring and summer. Historically, this maximum often occurs in late March or early April depending on the particular weather patterns in each year. 

The next cell has code showing how to find the maximum SWE value in a DataFrame (in this case, the data for 2025): 

```python
peakSWE_2025 = swe_2025["Snow Water Equivalent (in)"].max()
print("Peak Snow Water Equivalent in 2025:", peakSWE_2025, "inches")
```

The next cell has code showing how to find the date on which the maximum occurred by using the pandas `idxmax` function. `idxmax` returns the index (i.e. the row number or row name) of the maximum value. 

```python
peak_row_2025 = swe_2025["Snow Water Equivalent (in)"].idxmax()
print("Date of Peak Snow Water Equivalent in 2025:", swe_2025.loc[peak_row_2025, "Date"])
```
