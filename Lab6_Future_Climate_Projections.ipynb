**Data**

The climate data for this lab comes from global climate models that took part in the Coupled Model Intercomparison Project Phase 6 (CMIP6). The models simulate future climate change from 2015 to 2099 under different greenhouse gas emission scenarios, which are described later in the lab.

The data is stored in netcdf files. Each file contains the **annual mean temperature** (the average temperature over the whole year, in °C) for the climate model grid cell closest to Chicago, IL. The values are given as an **anomaly** compared to the 1985-2014 average (so a value of +3 means the year was 3°C warmer than the 1985-2014 average).

The file names have the format `temperature_Chicago_<model>_<scenario>_<simulation>.nc` and are stored at the following URL on the course github: 

```python
data_url = "https://raw.githubusercontent.com/fdavenport/CIVE480A-Fall26/refs/heads/main/data/"
```

The next cell downloads all of the files used in this lab:

```python
filenames = [
    "temperature_Chicago_CESM2_ssp370_r4i1p1f1.nc",
    "temperature_Chicago_MIROC6_ssp370_r1i1p1f1.nc",
    "temperature_Chicago_CanESM5_ssp370_r1i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp370_r1i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp370_r4i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp370_r5i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp126_r1i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp245_r1i1p1f1.nc",
    "temperature_Chicago_ACCESS-ESM1-5_ssp585_r1i1p1f1.nc",
]

for filename in filenames:
    if not os.path.exists(filename):
        urllib.request.urlretrieve(data_url + filename, filename)
```

**Future projections: natural variability**

Even if we had a perfect climate model and knew exactly what emissions would be, the climate would still vary from year to year and decade to decade because of natural processes (for example El Niño and La Niña, randomness in day-to-day weather patterns, etc). Because this variability is chaotic, climate scientists run the same model and scenario several times with slightly different initial conditions to capture the effect of this randomness. 

You will analyze three simulations from the ACCESS-ESM1-5 model using the same `ssp370` scenario. These simulations are labeled: `r1i1p1f1`, `r4i1p1f1`, and `r5i1p1f1`. (simulation `r1i1p1f1` was loaded as `access_ssp370_r1` above.)

We will focus on two future time periods that are relevant for planning: **mid-century** (2035-2064) and **end-of-century** (2070-2099). Because the anomaly is relative to 1985-2014, the average anomaly during each period is the **change in annual mean temperature** compared to the recent past. The next cell defines these periods:

```python
midcentury = slice(2035, 2064)
endcentury = slice(2070, 2099)
```

The next cell calculates the mid-century and end-of-century changes (the average annual mean temperature anomaly during each time period):

```python
access_r1_mid_change = access_ssp370_r1["meanT"].sel(year = midcentury).mean().item()
access_r1_end_change = access_ssp370_r1["meanT"].sel(year = endcentury).mean().item()
print("ACCESS-ESM1-5 r1 mid-century change: %.1f°C" % access_r1_mid_change)
print("ACCESS-ESM1-5 r1 end-of-century change: %.1f°C" % access_r1_end_change)
```
